```python
from z3 import Int, Distinct, And, Or, Solver, sat
```

```python
def check_solution(s):
    print("Solving the following conditions:")
    print(s)

    if s.check() == sat:
        print("Solution found:")
        print(s.model())
    else:
        print("No solution exists")
```

## Question 1

Four students—Alice, Bob, Charlie, and David—receive different integer exam scores between 0 and 100. Alice scores higher than Bob but lower than Charlie. David scores exactly 10 points more than Bob. Their average is 75, and Charlie scores at least 90. Find one possible set of scores.

```python
A = Int('A')  
B = Int('B')
C = Int('C')
D = Int('D')

s1 = Solver()

s1.add(Distinct(A, B, C, D))

s1.add(And(A >= 0, A <= 100))
s1.add(And(B >= 0, B <= 100))
s1.add(And(C >= 0, C <= 100))
s1.add(And(D >= 0, D <= 100)) 

s1.add(A > B)
s1.add(A < C)
s1.add(C >= 90)
s1.add(D == B + 10)
s1.add(A + B + C + D == 75 * 4)

check_solution(s1)
```

## Question 2

Four friends—Alice, Bob, Charlie, and David—sit in four chairs numbered 1 through 4. Each occupies exactly one chair, with no shared chairs. Alice sits to the left of Bob. Charlie is not in an end chair. David sits immediately next to Alice. Find a valid arrangement.

```python
A = Int('A')
B = Int('B')
C = Int('C')
D = Int('D')

s2 = Solver()

s2.add(And(A >= 1, A <= 4))
s2.add(And(B >= 1, B <= 4))
s2.add(And(C >= 1, C <= 4))
s2.add(And(D >= 1, D <= 4)) 

s2.add(Distinct(A, B, C, D))

s2.add(A < B)
s2.add(And(C != 1, C != 4))
s2.add(Or(D == A-1, D == A+1))

check_solution(s2)
```

## Question 3

A password has exactly four digits, each between 0 and 9. The first digit is not zero, and all four digits are different. The second digit is twice the first. The third digit is greater than the fourth. Their sum is 18. Find a valid password.

```python
D1 = Int('D1')
D2 = Int('D2')
D3 = Int('D3')
D4 = Int('D4')

s3 = Solver()

s3.add(Distinct(D1, D2, D3, D4))

s3.add(And(D1 >= 0, D1 <= 9))
s3.add(And(D2 >= 0, D2 <= 9))
s3.add(And(D3 >= 0, D3 <= 9))
s3.add(And(D4 >= 0, D4 <= 9))

s3.add(D1 != 0)
s3.add(D2 == D1 * 2)
s3.add(D3 > D4)
s3.add(D1 + D2 + D3 + D4 == 18)

check_solution(s3)
```

## Question 4

A delivery company must deliver packages A, B, C, and D on four different days, numbered 1 through 4. Exactly one package is delivered each day. A must be delivered before C. B must be delivered immediately after D. C cannot be delivered on the last day. Find a valid schedule.

```python
A = Int('A')
B = Int('B')
C = Int('C')
D = Int('D')

s4 = Solver()

s4.add(Distinct(A, B, C, D))

s4.add(And(A >= 1, A <= 4))
s4.add(And(B >= 1, B <= 4))
s4.add(And(C >= 1, C <= 4))
s4.add(And(D >= 1, D <= 4))

s4.add(A < C)
s4.add(B == D+1)
s4.add(C != 4)

check_solution(s4)
```

## Question 5

A customer buys exactly 10 items: notebooks, pens, and pencils. A notebook costs $5, a pen costs $2, and a pencil costs $1. The customer spends exactly $25, buys at least one of each item, and buys more pens than notebooks. Find a valid combination.

```python
Notebook = Int('Notebook')
Pen = Int('Pen')
Pencil = Int('Pencil')

s5 = Solver()

s5.add(Notebook + Pen + Pencil == 10)
s5.add(Notebook >= 1)
s5.add(Pen >= 1)
s5.add(Pencil >= 1)
s5.add(Notebook * 5 + Pen * 2 + Pencil * 1 == 25)
s5.add(Pen > Notebook)

check_solution(s5)
```
