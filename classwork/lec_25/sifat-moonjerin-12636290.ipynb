## Lecture 25

Place N queens on an N × N chessboard so that no two queens attack each other.
Express the puzzle as a satisfiability problem. Start with N = 4.

```python
from z3 import Int, Distinct, And, Solver, sat, Not
```

```python
N = 4

s = Solver()  
```

## Variables

```python
Q = [Int(f"Q_{i}") for i in range(N)] # Q[i] represents the column position of the queen in row i
```

```python
for i in range(N):
    s.add(And(Q[i] >= 0, Q[i] < N)) # Ensures the column position of each queen is between 0 and N-1
```

## Column Constraints

```python
s.add(Distinct(Q)) # Ensures that no two queens are in the same column
```

## Step 4: Diagonal Constraints

```python
s.add(Distinct(row - Q[row] for row in range(N))) 
s.add(Distinct(row + Q[row] for row in range(N)))

s
```

## Step 5: Solution

```python
count = 0
while s.check() == sat:
    count += 1
    m = s.model()
    print(f"Solution {count} found:")

    for i in range(N):
        print(f"Queen {i + 1}: Row {i}, Column {m[Q[i]]}")

    s.add(Not(And([Q[i] == m[Q[i]] for i in range(N)])))  # Avoids finding the same solution again
        
else:
    if count == 0:
        print("No solution exists")
    else:
        print(f"Total solutions found: {count}")
```
