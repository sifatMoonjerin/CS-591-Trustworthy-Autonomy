```python
from z3 import Bool, And, Or, Not, Implies, Solver, sat
```

```python
def check_solution(s):
    print("Solving the following conditions:")
    print(s)

    if s.check() == sat:
        print("Solution found:")
        print(s.model())
    else:
        print("No solution exists")
```

## Question 1

```python
A = Bool('A')
M = Bool('M')
R = Bool('R')

s1 = Solver()

s1.add(Implies(A, M))

s1.add(Not(And(M, R)))

s1.add(Or(A, R))

check_solution(s1)
```

## Question 2

```python
H = Bool('H')
S = Bool('S')
M = Bool('M')

s2 = Solver()

s2.add(Or(H, S, M))

s2.add(Implies(H, Not(S)))

s2.add(Implies(M, S))

s2.add(Not(And(H, S, M)))

check_solution(s2)
```

## Question 3

```python
A = Bool('A')
B = Bool('B')
C = Bool('C')
D = Bool('D')

s3 = Solver()

s3.add(Implies(A, B))

s3.add(Implies(C, Not(D)))

s3.add(Or(A, C))

s3.add(Not(And(B, D)))

check_solution(s3)
```

## Question 4

```python
L = Bool('L')
E = Bool('E')
B = Bool('B')
R = Bool('R')

s4 = Solver()

s4.add(Implies(R, E))

s4.add(Implies(B, L))

s4.add(Or(R, L))

s4.add(Not(And(E, B)))

check_solution(s4)
```

## Question 5

```python
A = Bool('A')
M = Bool('M')

s5 = Solver()

s5.add(Or(A, M))

s5.add(Not(And(A, M)))

s5.add(Implies(A, M))

s5.add(Implies(M, A))

check_solution(s5)
```
