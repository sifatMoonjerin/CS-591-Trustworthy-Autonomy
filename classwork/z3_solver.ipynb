```python
from z3 import Int, solve, simplify, And
```

```python
x = Int('x')
y = Int('y')

solve(x > 2, y < 10, x + 2*y == 7)
```

```python
x = Int('x')
y = Int('y')
print (simplify(x + y + 2*x + 3))
print (simplify(x < y + x + 2))
print (simplify(And(x + 1 >= 3, x**2 + x**2 + y**2 + 2 >= 5)))
```
