```python
from z3 import Bool, And, Or, Not, Xor, Implies, Solver, sat, simplify 
```

## Question 1

```python
A = Bool('A')
M = Bool('M')
R = Bool('R')

s = Solver()

condition_1 = Implies(A, M)

condition_2 = Xor(M, R)

condition_3 = Or(A, R)

s.add(condition_1, condition_2, condition_3)

print("Solving the following conditions:")
print(s)

if s.check() == sat:
    print("Solution found:")
    print(s.model())
else:
    print("No solution exists")
```

## Question 2

```python
H = Bool('H')
S = Bool('S')
M = Bool('M')

s = Solver()

condition_1 = Or(H, S, M)

condition_2 = Implies(H, Not(S))

condition_3 = Implies(M, S)

condition_4 = Not(And(H, S, M))

s.add(condition_1, condition_2, condition_3, condition_4)

print("Solving the following conditions:")
print(s)

if s.check() == sat:
    print("Solution found:")
    print(s.model())
else:
    print("No solution exists")
```
