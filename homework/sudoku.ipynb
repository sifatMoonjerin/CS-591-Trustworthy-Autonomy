```python
from z3 import Bool, And, Or, Not, Implies, Solver, sat, is_true
```

```python
n = 9

box_size = int(n ** 0.5)

```

```python
s = Solver()
```

```python
sudoku = {}

for row in range(n):
    for col in range(n):
        for k in range(1, n + 1):
            sudoku[(row, col, k)] = Bool(f"V_{row}{col}_{k}")
```

```python
for row in range(n):
    for col in range(n):
        s.add(
            Or(
                sudoku[(row, col, k)] for k in range(1, n + 1)
            )
        )

        for k in range(1, n + 1):
            s.add(
                Implies(
                    sudoku[(row, col, k)],
                    Not(Or(
                        sudoku[(r, col, k)] for r in range(n) if r != row
                    ))
                )
            )

            s.add(
                Implies(
                    sudoku[(row, col, k)],
                    Not(Or(
                        sudoku[(row, c, k)] for c in range(n) if c != col
                    ))
                )
            )

            box_row = (row // box_size) * box_size
            box_col = (col // box_size) * box_size

            s.add(
                Implies(
                    sudoku[(row, col, k)],
                    Not(Or(
                        sudoku[(r, c, k)] 
                        for r in range(box_row, box_row + 3) 
                        for c in range(box_col, box_col + 3) 
                        if not(r == row and c == col)
                    ))
                )
            )


```

```python
s.add(
    sudoku[(0, 1, 1)],
    sudoku[(0, 3, 4)],
    sudoku[(0, 5, 2)],
    sudoku[(0, 7, 5)],
    sudoku[(1, 8, 6)],
    sudoku[(2, 3, 3)],
    sudoku[(2, 5, 1)],
    sudoku[(3, 0, 7)],
    sudoku[(3, 2, 5)],
    sudoku[(3, 6, 4)],
    sudoku[(3, 8, 8)],
    sudoku[(5, 0, 2)],
    sudoku[(5, 2, 8)],
    sudoku[(5, 6, 5)],
    sudoku[(5, 8, 9)],
    sudoku[(6, 3, 9)],
    sudoku[(6, 5, 6)],
    sudoku[(7, 0, 6)],
    sudoku[(7, 8, 2)],
    sudoku[(8, 1, 7)],
    sudoku[(8, 3, 1)],
    sudoku[(8, 5, 3)],
    sudoku[(8, 7, 4)]
)
```

```python
print(s.check())
```

```python
if s.check() == sat:
    print("Solution found:")
    model = s.model()
    
    for var in model:
        if is_true(model[var]):
            print(var)
else:
    print("No solution exists")
```
