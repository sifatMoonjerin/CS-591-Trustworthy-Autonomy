## Embedding Color Variables

```python
colors = {}

for v in vertices:

    colors[(v, 'r')] = Bool(f"V_{v}_r")
    colors[(v, 'g')] = Bool(f"V_{v}_g")
    colors[(v, 'b')] = Bool(f"V_{v}_b")
    colors[(v, 'y')] = Bool(f"V_{v}_y")
```

```python
s = Solver()
```

## Adding Contraints

```python
for v in vertices:
    s.add(
        Or(
            colors[(v, 'r')], 
            colors[(v, 'g')], 
            colors[(v, 'b')], 
            colors[(v, 'y')]
        )
    ) # Each vertex has at least one color

    s.add(
        And(
            Or(Not(colors[(v, 'r')]), Not(colors[(v, 'g')])),
            Or(Not(colors[(v, 'r')]), Not(colors[(v, 'b')])),
            Or(Not(colors[(v, 'r')]), Not(colors[(v, 'y')])),
            Or(Not(colors[(v, 'g')]), Not(colors[(v, 'b')])),
            Or(Not(colors[(v, 'g')]), Not(colors[(v, 'y')])),
            Or(Not(colors[(v, 'b')]), Not(colors[(v, 'y')])),
        )
    ) # Each vertex has only one color


for u, v in edges:
    s.add(
        And(
            Or(Not(colors[(u, 'r')]), Not(colors[(v, 'r')])),
            Or(Not(colors[(u, 'g')]), Not(colors[(v, 'g')])),
            Or(Not(colors[(u, 'b')]), Not(colors[(v, 'b')])),
            Or(Not(colors[(u, 'y')]), Not(colors[(v, 'y')])),
        )
    ) # connected nodes do not have the same color
```

```python
print("Solving the following conditions:")
print(s)
```

```python
color_map = {
    'r': 'Red',
    'g': 'Green',
    'b': 'Blue',
    'y': 'Yellow'
}

if s.check() == sat:
    print("Solution found:")
    model = s.model()
    for v in vertices:
        for c in ['r', 'g', 'b', 'y']:
            if model.evaluate(colors[(v, c)]):
                print(f"{v} = {color_map[c]}")
else:
    print("No solution exists")
```
